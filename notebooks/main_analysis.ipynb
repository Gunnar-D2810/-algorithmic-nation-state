# Algorithmic Nation-State Main Analysis

This notebook runs and summarizes the integrated research workflow. It keeps empirical forecasting diagnostics, exploratory ABM simulations, and subjective probabilistic updates separate.

```python
from pathlib import Path
import subprocess
import sys
import pandas as pd
from IPython.display import Image, Markdown, display

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
TABLES = PROJECT_ROOT / "reports" / "tables"
FIGURES = PROJECT_ROOT / "reports" / "figures"
print(PROJECT_ROOT)
```

## Run The Complete Workflow

The pipeline validates existing ingested data by default. It does not refresh World Bank data unless `--refresh-data` is supplied.

```python
subprocess.run([
    sys.executable,
    str(PROJECT_ROOT / "scripts" / "run_full_pipeline.py"),
    "--monte-carlo-iterations", "100",
], cwd=PROJECT_ROOT, check=True)
```

## Empirical Evidence

These tables summarize observed macroeconomic panel coverage and forecasting diagnostics. They should not be interpreted as causal estimates.

```python
macro = pd.read_csv(PROJECT_ROOT / "data" / "processed" / "macro_panel.csv")
coverage = macro.groupby("country")["value"].agg(rows="size", observed=lambda s: int(s.notna().sum()))
display(coverage)
display(pd.read_csv(TABLES / "all_model_comparison.csv").sort_values(["rmse", "model"]).head(12))
```

## Forecasting And Interpretability Figures

```python
for figure in [
    FIGURES / "integrated" / "publication_forecasting_comparison.png",
    FIGURES / "integrated" / "publication_feature_importance.png",
]:
    if figure.exists():
        display(Image(filename=str(figure)))
```

## Exploratory Simulations

ABM outputs are simulated diagnostics conditional on conceptual assumptions, not geopolitical predictions.

```python
display(pd.read_csv(TABLES / "abm" / "abm_scenario_comparison.csv"))
figure = FIGURES / "integrated" / "publication_abm_scenario_trajectories.png"
if figure.exists():
    display(Image(filename=str(figure)))
```

## Probabilistic Stress Testing

Bayesian posteriors use weak priors and simulation evidence proxies. Monte Carlo outputs propagate uncertainty through stated ABM perturbations.

```python
display(pd.read_csv(TABLES / "probabilistic" / "bayesian_posteriors.csv").head(12))
display(pd.read_csv(TABLES / "probabilistic" / "monte_carlo_summary.csv").head(12))
for figure in [
    FIGURES / "integrated" / "publication_bayesian_posteriors.png",
    FIGURES / "integrated" / "publication_monte_carlo_uncertainty_bands.png",
    FIGURES / "integrated" / "integrated_summary_dashboard.png",
]:
    if figure.exists():
        display(Image(filename=str(figure)))
```
