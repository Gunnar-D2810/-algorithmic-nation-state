# Forecasting Analysis

This notebook reruns baseline and modern GDP-growth forecasting benchmarks and displays conservative diagnostic outputs.

```python
from pathlib import Path
import subprocess
import sys
import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
TABLES = PROJECT_ROOT / "reports" / "tables"
FIGURES = PROJECT_ROOT / "reports" / "figures"
```

```python
subprocess.run([sys.executable, str(PROJECT_ROOT / "scripts" / "run_forecasting.py")], cwd=PROJECT_ROOT, check=True)
subprocess.run([sys.executable, str(PROJECT_ROOT / "scripts" / "run_modern_forecasting.py")], cwd=PROJECT_ROOT, check=True)
```

## Model Comparison

```python
comparison = pd.read_csv(TABLES / "all_model_comparison.csv")
display(comparison.sort_values(["country", "rmse", "model"]).head(20))
figure = FIGURES / "model_rankings.png"
if figure.exists():
    display(Image(filename=str(figure)))
```
