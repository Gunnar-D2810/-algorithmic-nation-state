# ECAIF ABM Analysis

This notebook reruns the exploratory Evolutionary Cloud AI Firms ABM and displays scenario diagnostics.

```python
from pathlib import Path
import subprocess
import sys
import pandas as pd
from IPython.display import Image, Markdown, display

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
TABLES = PROJECT_ROOT / "reports" / "tables"
FIGURES = PROJECT_ROOT / "reports" / "figures"
```

```python
subprocess.run([sys.executable, str(PROJECT_ROOT / "scripts" / "run_abm.py"), "--seed", "42", "--time-steps", "30"], cwd=PROJECT_ROOT, check=True)
```

```python
display(pd.read_csv(TABLES / "abm" / "abm_scenario_comparison.csv"))
display(pd.read_csv(TABLES / "abm" / "abm_reproducibility_check.csv"))
for figure in [
    FIGURES / "abm" / "scenario_comparison.png",
    FIGURES / "abm" / "ai_capability_divergence.png",
    FIGURES / "abm" / "dependency_score.png",
]:
    if figure.exists():
        display(Image(filename=str(figure)))
```
